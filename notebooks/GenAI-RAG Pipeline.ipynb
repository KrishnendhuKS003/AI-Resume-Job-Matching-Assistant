PDF extraction function

```python
def extract_text_from_pdf(file_path):
    """
    Extract text from a PDF file.
    """
    document = pymupdf.open(file_path)

    pages = []

    for page in document:
        pages.append(page.get_text())

    document.close()

    return "\n".join(pages).strip()
```

DOCX extraction function

```python
def extract_text_from_docx(file_path):
    """
    Extract text from a DOCX file.
    """
    document = Document(file_path)

    paragraphs = [
        paragraph.text
        for paragraph in document.paragraphs
        if paragraph.text.strip()
    ]

    return "\n".join(paragraphs).strip()
```

Generic document extractor

```python
def extract_document_text(file_path):
    """
    Extract text from PDF or DOCX.
    """
    file_path = Path(file_path)

    suffix = file_path.suffix.lower()

    if suffix == ".pdf":
        return extract_text_from_pdf(file_path)

    elif suffix == ".docx":
        return extract_text_from_docx(file_path)

    else:
        raise ValueError(
            f"Unsupported file type: {suffix}. "
            "Use PDF or DOCX."
        )
```

Resume section extraction

```python
SECTION_PATTERNS = {
    "summary": [
        "summary",
        "professional summary",
        "profile",
        "objective",
        "career objective"
    ],
    "skills": [
        "skills",
        "technical skills",
        "core skills",
        "key skills"
    ],
    "experience": [
        "experience",
        "work experience",
        "professional experience",
        "employment"
    ],
    "education": [
        "education",
        "academic background",
        "qualifications"
    ],
    "projects": [
        "projects",
        "academic projects",
        "personal projects"
    ],
    "certifications": [
        "certifications",
        "certificates",
        "courses"
    ]
}

print("Section patterns created.")
```

Section detection helper

```python
def detect_section_heading(line):
    """
    Detect whether a line looks like a resume section heading.
    """
    normalized = re.sub(r"[^a-zA-Z ]", "", line.lower()).strip()

    for section, headings in SECTION_PATTERNS.items():
        if normalized in headings:
            return section

    return None
```

Resume section parser

```python
def extract_resume_sections(text):
    """
    Split resume text into basic sections.
    """
    lines = text.splitlines()

    sections = {}
    current_section = "general"

    sections[current_section] = []

    for line in lines:
        line = line.strip()

        if not line:
            continue

        detected = detect_section_heading(line)

        if detected:
            current_section = detected

            if current_section not in sections:
                sections[current_section] = []

        else:
            sections.setdefault(current_section, [])
            sections[current_section].append(line)

    # Convert lists to strings
    sections = {
        section: "\n".join(content).strip()
        for section, content in sections.items()
    }

    return sections
```

```python
# Cell 32 — Sentence-aware document chunking function

import re

def chunk_text(text, chunk_size=500, overlap=100):
    """
    Split text into chunks while trying to preserve complete sentences.

    Short texts are kept as a single chunk.
    Longer texts are grouped into chunks close to the target size.
    """

    text = str(text).strip()

    if not text:
        return []

    # Keep short documents as one chunk
    if len(text) <= chunk_size:
        return [text]

    # Split text into sentences while preserving sentence-ending punctuation
    sentences = re.split(r'(?<=[.!?])\s+', text)

    chunks = []
    current_chunk = ""

    for sentence in sentences:

        sentence = sentence.strip()

        if not sentence:
            continue

        # If adding the sentence stays within the target size
        if len(current_chunk) + len(sentence) + 1 <= chunk_size:
            current_chunk += (" " if current_chunk else "") + sentence

        else:
            # Save the current chunk
            if current_chunk:
                chunks.append(current_chunk.strip())

            # Start a new chunk with the current sentence
            current_chunk = sentence

    # Add the final chunk
    if current_chunk:
        chunks.append(current_chunk.strip())

    return chunks
```

```python
# Cell 48B — Convert skill fields back to proper Python lists

import ast

def ensure_list(value):
    """Convert a list-like string into a Python list."""
    
    if isinstance(value, list):
        return value
    
    if isinstance(value, str):
        try:
            parsed = ast.literal_eval(value)
            
            if isinstance(parsed, list):
                return parsed
            
        except (ValueError, SyntaxError):
            pass
        
        # Fallback: treat the string as one item
        return [value]
    
    return [value]


# Fix the fields
final_result["matching_skills"] = ensure_list(
    final_result["matching_skills"]
)

final_result["missing_skills"] = ensure_list(
    final_result["missing_skills"]
)

final_result["relevant_experience"] = ensure_list(
    final_result["relevant_experience"]
)


# Verify
print("Matching skills type:",
      type(final_result["matching_skills"]))

print("Missing skills type:",
      type(final_result["missing_skills"]))

print("Relevant experience type:",
      type(final_result["relevant_experience"]))

print("\nMatching skills:")
print(final_result["matching_skills"])

print("\nMissing skills:")
print(final_result["missing_skills"])
```

```python
# Cell 33 — Create embeddings for resume and JD chunks

from sentence_transformers import SentenceTransformer
import numpy as np

embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

resume_embeddings = embedding_model.encode(
    resume_chunks,
    convert_to_numpy=True,
    normalize_embeddings=True
)

jd_embeddings = embedding_model.encode(
    jd_chunks,
    convert_to_numpy=True,
    normalize_embeddings=True
)

print("Resume embeddings shape:", resume_embeddings.shape)
print("JD embeddings shape:", jd_embeddings.shape)
```

```python
# Cell 34 — Create FAISS index for resume chunks

import faiss
import numpy as np

# Number of dimensions in our embeddings
embedding_dimension = resume_embeddings.shape[1]

# Create FAISS index using inner product
# Because our embeddings are normalized, inner product = cosine similarity
resume_index = faiss.IndexFlatIP(embedding_dimension)

# Add resume chunk embeddings to the index
resume_index.add(resume_embeddings.astype("float32"))

print("FAISS index created successfully.")
print("Embedding dimension:", embedding_dimension)
print("Number of resume chunks in index:", resume_index.ntotal)
```

```python
# Cell 35 — Retrieve the most relevant resume chunks

# Number of chunks to retrieve
top_k = min(3, len(resume_chunks))

# Search FAISS using the job description embedding
similarities, indices = resume_index.search(
    jd_embeddings.astype("float32"),
    top_k
)

print("Top matching resume chunks:\n")

for rank, (score, index) in enumerate(
    zip(similarities[0], indices[0]), start=1
):
    print(f"Rank {rank}")
    print(f"Similarity score: {score:.4f}")
    print(f"Resume chunk index: {index}")
    print(f"Text: {resume_chunks[index]}")
    print("-" * 80)
```

```python
# Cell 36 — Build retrieved context

retrieved_chunks = []

for index in indices[0]:
    retrieved_chunks.append(resume_chunks[index])

retrieved_context = "\n\n".join(retrieved_chunks)

print("Retrieved context created successfully.")
print("\nNumber of retrieved chunks:", len(retrieved_chunks))

print("\nRetrieved context:\n")
print(retrieved_context)
```
